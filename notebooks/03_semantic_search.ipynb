```python
from sentence_transformers import SentenceTransformer

print("Sentence Transformers imported successfully")
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
documents = [
    "Machine learning allows computers to learn patterns from data.",
    "Artificial intelligence enables machines to perform tasks that normally require human intelligence.",
    "Natural language processing helps computers understand and process human language.",
    "Reinforcement learning allows an agent to learn by interacting with an environment.",
    "Statistics helps us analyse data and understand uncertainty."
]
```

```python
document_embeddings = model.encode(documents)

print(document_embeddings.shape)
```

```python
print(document_embeddings[0])
```

```python
query = "How do computers learn from data?"
query_embedding = model.encode([query])

print(query_embedding.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarities = cosine_similarity(
    query_embedding,
    document_embeddings
)

print(similarities)
```

```python
scores = similarities[0]

for i, score in enumerate(scores):
    print(f"Document {i}: {score:.4f}")
```

```python
ranked_indices = scores.argsort()[::-1]

for rank, index in enumerate(ranked_indices, start=1):
    print(f"Rank {rank}")
    print(f"Score: {scores[index]:.4f}")
    print(f"Document: {documents[index]}")
    print()
```

Now test the important semantic example

```python
query = "How do machines acquire knowledge from datasets?"
query_embedding = model.encode([query])
similarities = cosine_similarity(
    query_embedding,
    document_embeddings
)[0]
```

```python
ranked_indices = similarities.argsort()[::-1]

for rank, index in enumerate(ranked_indices, start=1):
    print(f"Rank {rank}")
    print(f"Score: {similarities[index]:.4f}")
    print(f"Document: {documents[index]}")
    print()
```

```python
def semantic_search(query, documents, model, document_embeddings, top_k=3):

    query_embedding = model.encode([query])

    similarities = cosine_similarity(
        query_embedding,
        document_embeddings
    )[0]

    ranked_indices = similarities.argsort()[::-1]

    results = []

    for index in ranked_indices[:top_k]:
        results.append({
            "document": documents[index],
            "score": similarities[index]
        })

    return results
```

```python
results = semantic_search(
    "How do machines acquire knowledge from datasets?",
    documents,
    model,
    document_embeddings,
    top_k=3
)

for result in results:
    print(f"{result['score']:.4f} | {result['document']}")
```

```python
queries = [
    "How do computers learn from data?",
    "How does an agent learn?",
    "What helps computers understand human language?",
    "How do we analyse uncertainty?"
]

for query in queries:

    print(f"\nQuery: {query}")
    print("-" * 60)

    results = semantic_search(
        query,
        documents,
        model,
        document_embeddings,
        top_k=2
    )

    for result in results:
        print(f"{result['score']:.4f} | {result['document']}")
```

```python
semantic_queries = [
    "How do machines acquire knowledge from datasets?",
    "How can a computer understand human speech and text?",
    "How does an intelligent system learn through interaction?",
    "How can we measure uncertainty in information?"
]

for query in semantic_queries:

    print(f"\nQuery: {query}")
    print("-" * 60)

    results = semantic_search(
        query,
        documents,
        model,
        document_embeddings,
        top_k=2
    )

    for result in results:
        print(f"{result['score']:.4f} | {result['document']}")
```

```python
evaluation_data = [
    {
        "query": "How do computers learn from data?",
        "relevant_document": 0
    },
    {
        "query": "How does an agent learn?",
        "relevant_document": 3
    },
    {
        "query": "What helps computers understand human language?",
        "relevant_document": 2
    },
    {
        "query": "How do we analyse uncertainty?",
        "relevant_document": 4
    }
]
```

```python
def semantic_hit_at_k(query, relevant_document, k):

    results = semantic_search(
        query,
        documents,
        model,
        document_embeddings,
        top_k=k
    )

    retrieved_documents = []

    for result in results:
        retrieved_documents.append(result["document"])

    relevant_text = documents[relevant_document]

    return relevant_text in retrieved_documents
```

```python
hits = 0

for item in evaluation_data:

    result = semantic_hit_at_k(
        item["query"],
        item["relevant_document"],
        k=2
    )

    if result:
        hits += 1

hit_at_2 = hits / len(evaluation_data)

print(f"Semantic Search Hit@2: {hit_at_2:.2f}")
```

```python
hits = 0

for item in evaluation_data:

    result = semantic_hit_at_k(
        item["query"],
        item["relevant_document"],
        k=1
    )

    if result:
        hits += 1

hit_at_1 = hits / len(evaluation_data)

print(f"Semantic Search Hit@1: {hit_at_1:.2f}")
```

## Retrieval Method Comparison

| Method | Main idea | Type |
|---|---|---|
| TF-IDF | Measures importance of words | Lexical |
| BM25 | Improves lexical relevance scoring | Lexical |
| Semantic Search | Compares meaning using embeddings | Semantic |

All three methods were evaluated using the same retrieval task.

## Limitations of Semantic Search

Semantic search can identify similarity in meaning even when
different words are used. However, it also has limitations.

The quality of retrieval depends on the embedding model.
Semantically similar documents may not always be relevant to
the specific question.

Embedding-based retrieval can also miss exact identifiers,
rare terms, names, numbers, or technical keywords that lexical
retrieval can handle well.

Therefore, semantic search should not automatically replace
lexical retrieval. This motivates combining lexical and semantic
retrieval in a hybrid system.
