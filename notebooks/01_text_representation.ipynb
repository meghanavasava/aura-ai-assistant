```python
documents = [
    "Machine learning allows computers to learn patterns from data.",
    "Artificial intelligence enables machines to perform tasks that normally require human intelligence.",
    "Natural language processing helps computers understand and process human language.",
    "Reinforcement learning allows an agent to learn by interacting with an environment.",
    "Statistics helps us analyse data and understand uncertainty."
]

for i, document in enumerate(documents):
    print(f"Document {i}: {document}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(documents)

print("Shape:", X.shape)
```

```python
vocabulary = vectorizer.get_feature_names_out()

print(vocabulary)
```

```python
import pandas as pd

tfidf_df = pd.DataFrame(
    X.toarray(),
    columns=vocabulary
)

tfidf_df
```

```python
query = "How can an agent learn through interaction?"

query_vector = vectorizer.transform([query])

print(query_vector.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarities = cosine_similarity(query_vector, X)

print(similarities)
```

```python
scores = similarities[0]

for i, score in enumerate(scores):
    print(f"Document {i}: {score:.4f}")
```

```python
ranked_indices = scores.argsort()[::-1]

for rank, index in enumerate(ranked_indices, start=1):
    print(f"Rank {rank}")
    print(f"Score: {scores[index]:.4f}")
    print(f"Document: {documents[index]}")
    print()
```

```python
def search(query, documents, vectorizer, document_vectors, top_k=3):
    query_vector = vectorizer.transform([query])

    similarities = cosine_similarity(
        query_vector,
        document_vectors
    )[0]

    ranked_indices = similarities.argsort()[::-1]

    results = []

    for index in ranked_indices[:top_k]:
        results.append({
            "document": documents[index],
            "score": similarities[index]
        })

    return results
```

```python
results = search(
    "What allows computers to gain knowledge from examples?",
    documents,
    vectorizer,
    X
)

for result in results:
    print(result)
```

```python
queries = [
    "How do computers learn from data?",
    "How does an agent learn?",
    "What helps computers understand human language?",
    "How do we analyse uncertainty?"
]

for query in queries:
    print(f"\nQuery: {query}")
    print("-" * 60)

    results = search(
        query,
        documents,
        vectorizer,
        X,
        top_k=2
    )

    for result in results:
        print(f"{result['score']:.4f} | {result['document']}")
```

```python
evaluation_data = [
    {
        "query": "How do computers learn from data?",
        "relevant_document": 0
    },
    {
        "query": "How does an agent learn?",
        "relevant_document": 3
    },
    {
        "query": "What helps computers understand human language?",
        "relevant_document": 2
    },
    {
        "query": "How do we analyse uncertainty?",
        "relevant_document": 4
    }
]
```

```python
def hit_at_k(query, relevant_document, k):
    results = search(
        query,
        documents,
        vectorizer,
        X,
        top_k=k
    )

    retrieved_documents = []

    for result in results:
        retrieved_documents.append(result["document"])

    relevant_text = documents[relevant_document]

    return relevant_text in retrieved_documents
```

```python
result = hit_at_k(
    "How do computers learn from data?",
    0,
    2
)

print(result)
```

```python
for item in evaluation_data:
    result = hit_at_k(
        item["query"],
        item["relevant_document"],
        k=2
    )

    print(f"Query: {item['query']}")
    print(f"Hit@2: {result}")
    print()
```

```python
hits = 0

for item in evaluation_data:
    result = hit_at_k(
        item["query"],
        item["relevant_document"],
        k=1
    )

    if result:
        hits += 1

hit_at_2 = hits / len(evaluation_data)

print(f"Hit@2: {hit_at_2:.2f}")
```

## Limitations of TF-IDF

TF-IDF retrieval works well when the query and document share
important words. However, TF-IDF mainly relies on lexical overlap
and does not understand the semantic meaning of words.

For example:

Query:
"How do computers learn from datasets?"

Document:
"Machine learning allows systems to acquire patterns from data."

These sentences have similar meanings, but they may receive a
low similarity score because many words are different.

This limitation motivates the use of more advanced retrieval
methods such as BM25 and semantic embeddings.

## Conclusion

In this experiment, I implemented a basic information retrieval
system using TF-IDF and cosine similarity.

The system:
- represents documents numerically using TF-IDF
- converts user queries into the same vector space
- calculates cosine similarity between queries and documents
- ranks documents according to similarity
- evaluates retrieval using Hit@K

The experiment also demonstrated an important limitation:
TF-IDF depends heavily on lexical overlap and does not capture
semantic similarity.

The next experiment will investigate BM25 as an improved
classical retrieval method.
